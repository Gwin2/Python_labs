Напишите программу с использованием декоратора, которая переопределяет функцию print() так, чтобы она печатала весь текст в верхнем регистре.

```python
# Декоратор для переопределения print()
def uppercase_print(func):
    """
    Декоратор, который модифицирует функцию print(),
    преобразуя весь текст в верхний регистр.
    """
    def wrapper(*args, **kwargs):
        # Преобразуем все строковые аргументы в верхний регистр
        new_args = [str(arg).upper() for arg in args]
        new_kwargs = {k: str(v).upper() for k, v in kwargs.items()}
        
        # Вызываем оригинальную функцию print с измененными аргументами
        func(*new_args, **new_kwargs)
    
    return wrapper


# Переопределяем стандартную функцию print()
print = uppercase_print(print)


# Пример использования
if __name__ == "__main__":
    print("hello world!")  # Вывод: HELLO WORLD!
    print("привет", "мир")  # Вывод: ПРИВЕТ МИР
    print("test", sep="-", end="!a\n")  # Вывод: TEST!
```
