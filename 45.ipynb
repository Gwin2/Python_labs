Будем называть словом любую последовательность из одной или более латинских букв.
Реализуйте класс Word, описывающий расширяемый набор слов. При создании экземпляра класс должен принимать один аргумент:
words — список, определяющий начальный набор слов. Если не передан, начальный набор слов считается пустым
Экземпляр класса Word должен иметь один атрибут:
words — список, содержащий набор слов
Класс Word должен иметь четыре метода экземпляра:
add_word() — метод, принимающий в качестве аргумента слово и добавляющий его в набор. Если слово уже есть в наборе, метод ничего не должен делать;
words_with_length() — метод, принимающий в качестве аргумента натуральное число n и возвращающий список слов из набора, длина которых равна n;
only() — метод, принимающий произвольное количество аргументов — букв, и возвращающий все слова из набора, которые включают в себя только переданные буквы;
avoid() — метод, принимающий произвольное количество аргументов — букв, и возвращающий все слова из списка words, которые не содержат ни одну из этих букв;
Слова в списках, возвращаемых методами words_with_length(), only() и avoid(), должны располагаться в том порядке, в котором они были добавлены.


```python
class Word:
    # Конструктор класса, принимает начальный набор слов (по умолчанию пустой список)
    def __init__(self, words=None):
        self.words = [] if words is None else words  # Набор слов

    # Метод add_word() для добавления слова в набор
    def add_word(self, word):
        if word not in self.words:  # Проверяем, что слово еще не в наборе
            self.words.append(word)

    # Метод words_with_length() для получения слов определенной длины
    def words_with_length(self, n):
        return [word for word in self.words if len(word) == n]

    # Метод only() для получения слов, состоящих только из указанных букв
    def only(self, *letters):
        allowed_letters = set(letters)  # Множество разрешенных букв
        return [word for word in self.words if set(word).issubset(allowed_letters)]

    # Метод avoid() для получения слов, которые не содержат указанные буквы
    def avoid(self, *letters):
        forbidden_letters = set(letters)  # Множество запрещенных букв
        return [word for word in self.words if not set(word).intersection(forbidden_letters)]


# Пример использования класса Word
if __name__ == "__main__":
    # Создание экземпляра класса Word с начальным набором слов
    word_processor = Word(["apple", "banana", "cherry", "date"])

    # Добавление нового слова
    word_processor.add_word("elderberry")
    word_processor.add_word("apple")  # Повторное добавление слова (ничего не делает)

    # Получение слов определенной длины
    print("Слова длиной 5:", word_processor.words_with_length(5))

    # Получение слов, состоящих только из указанных букв
    print("Слова только из букв 'a', 'p', 'l', 'e':", word_processor.only('a', 'p', 'l', 'e'))

    # Получение слов, которые не содержат указанные буквы
    print("Слова без букв 'a', 'e':", word_processor.avoid('a', 'e'))
```
