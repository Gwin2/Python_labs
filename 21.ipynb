Даны некоторые список целых чисел и число k его возможных разбиений. Требуется составить все возможные разбиения данного списка при данном k, при каждом отдельном разбиении найти сумму p элементов на границах подгрупп списка, полученных при данном разбиении (т. е. сумму первого и последнего элементов подгруппы). Найти число p для каждого возможного разбиения. Затем отобрать из этих чисел максимальное pmax и минимальное pmin и найти их разность pmax -  pmin.

```python
from itertools import combinations

# Функция для генерации всех возможных разбиений списка на k подгрупп
def generate_partitions(lst, k):
    n = len(lst)
    if k > n or k <= 0:
        return []  # Невозможно разбить список на k подгрупп
    
    partitions = []
    
    # Генерация индексов точек разделения
    for dividers in combinations(range(1, n), k - 1):
        current_partition = []
        prev_index = 0
        
        # Создание подгрупп на основе точек разделения
        for divider in dividers:
            current_partition.append(lst[prev_index:divider])
            prev_index = divider
        current_partition.append(lst[prev_index:])  # Добавляем последнюю подгруппу
        
        partitions.append(current_partition)
    
    return partitions


# Функция для вычисления суммы p для одного разбиения
def calculate_p(partition):
    p_sum = 0
    for group in partition:
        if len(group) > 0:
            p_sum += group[0] + group[-1]  # Суммируем первый и последний элементы подгруппы
    return p_sum


# Основная функция
def find_p_diff(lst, k):
    # Шаг 1: Генерация всех возможных разбиений
    partitions = generate_partitions(lst, k)
    
    if not partitions:
        return None  # Если невозможно сгенерировать разбиения
    
    # Шаг 2: Вычисление p для каждого разбиения
    p_values = [calculate_p(partition) for partition in partitions]
    
    # Шаг 3: Нахождение p_max и p_min
    p_max = max(p_values)
    p_min = min(p_values)
    
    # Шаг 4: Вычисление разности p_max - p_min
    result = p_max - p_min
    
    return result


# Пример использования
if __name__ == "__main__":
    lst = list(map(int, input("Введите список чисел через пробел: ").split()))
    k = int(input("Введите число k (число подгрупп): "))
    
    result = find_p_diff(lst, k)
    if result is not None:
        print(f"Разница между максимальным и минимальным p: {result}")
    else:
        print("Невозможно выполнить разбиение.")
```
