Реализуйте декоратор sandwich, который выводит тексты:

---- Верхний ломтик хлеба ----

---- Нижний ломтик хлеба ----

до и после вызова декорируемой функции соответственно.

```python
def sandwich(func):
    """
    Декоратор, который добавляет тексты до и после вызова функции.
    """
    def wrapper(*args, **kwargs):
        # Выводим текст перед вызовом функции
        print("---- Верхний ломтик хлеба ----")
        
        # Вызываем декорируемую функцию
        result = func(*args, **kwargs)
        
        # Выводим текст после вызова функции
        print("---- Нижний ломтик хлеба ----")
        
        return result  # Возвращаем результат вызова функции
    
    return wrapper


# Пример использования
if __name__ == "__main__":
    @sandwich
    def my_function():
        print("Это содержимое сэндвича!")

    # Вызываем декорированную функцию
    my_function()
```
