Реализуйте генераторную функцию primes(), которая принимает два аргумента в следующем порядке:

left — натуральное число;

right — натуральное число.

Функция должна возвращать генератор, порождающий последовательность простых чисел от left до right включительно, а затем возбуждающий исключение StopIteration.

```python
def is_prime(n):
    """Вспомогательная функция для проверки, является ли число простым."""
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):  # Проверяем делители до корня из n
        if n % i == 0:
            return False
    return True

def primes(left, right):
    """Генераторная функция для порождения простых чисел в диапазоне [left, right]."""
    for num in range(left, right + 1):  # Проходим по всем числам в диапазоне
        if is_prime(num):  # Проверяем, является ли число простым
            yield num  # Возвращаем простое число


# Пример использования
if __name__ == "__main__":
    # Генерация простых чисел в диапазоне [10, 30]
    generator = primes(10, 30)
    print("Простые числа в диапазоне [10, 30]:")
    for prime in generator:
        print(prime, end=" ")
```
