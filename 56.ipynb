Реализуйте генераторную функцию simple_sequence(), которая не принимает никаких аргументов.

Функция должна возвращать генератор, порождающий бесконечную возрастающую последовательность натуральных чисел, в которой каждое число встречается столько раз, каково оно: 1,2,2,3,3,3,4,4,4,4,..

```python
def simple_sequence():
    n = 1  # Начинаем с первого натурального числа
    while True:  # Бесконечный цикл
        for _ in range(n):  # Повторяем число n раз
            yield n  # Возвращаем текущее число
        n += 1  # Переходим к следующему натуральному числу


# Пример использования
if __name__ == "__main__":
    # Создаем генератор
    generator = simple_sequence()

    # Получаем первые 10 элементов последовательности
    for _ in range(10):
        print(next(generator), end=" ")
```
