Создайте список, состоящий из четных элементов списка из первой задачи (через циклы и генераторы)

```python
# Предыдущий список чисел от 0 до 100
previous_list = list(range(101))

# Создание списка четных элементов через генератор списков
even_numbers = [num for num in previous_list if num % 2 == 0]

print(even_numbers)
```
