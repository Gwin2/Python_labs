Реализуйте декоратор do_twice, вызывающий декорируемую функцию два раза.

```python
def do_twice(func):
    """
    Декоратор, который вызывает декорируемую функцию два раза.
    """
    def wrapper(*args, **kwargs):
        # Первый вызов функции
        result1 = func(*args, **kwargs)
        
        # Второй вызов функции
        result2 = func(*args, **kwargs)
        
        # Возвращаем результат второго вызова (или можно вернуть оба результата)
        return result2
    
    return wrapper


# Пример использования
if __name__ == "__main__":
    @do_twice
    def greet(name):
        print(f"Hello, {name}!")
        return f"Hello, {name}!"

    # Вызываем декорированную функцию
    result = greet("Alice")
    print("Result:", result)
```
