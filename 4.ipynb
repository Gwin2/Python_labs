Реализуйте функцию проверки числа на простоту. Принимает на входе натуральное число, на выходе даёт булево значение. Не забудьте сделать проверку на дурака!

```python
def is_prime(n):
    # Проверка на дурака: число должно быть натуральным (> 0)
    if not isinstance(n, int) or n <= 1:
        print("Ошибка: Введите натуральное число больше 1.")
        return False

    # Простые числа меньше или равные 3
    if n <= 3:
        return True

    # Если число делится на 2 или 3, оно не является простым
    if n % 2 == 0 or n % 3 == 0:
        return False

    # Проверяем делители от 5 до sqrt(n), пропуская чётные числа
    i = 5
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6

    return True

# Тестирование функции
if __name__ == "__main__":
    try:
        number = int(input("Введите натуральное число для проверки на простоту: "))
        if is_prime(number):
            print(f"{number} — простое число.")
        else:
            print(f"{number} — не простое число.")
    except ValueError:
        print("Ошибка: Введите корректное натуральное число.")
```
