Реализуйте генераторную функцию around(), которая принимает один аргумент:

iterable — итерируемый объект.

Функция должна возвращать генератор, порождающий последовательность кортежей, каждый из которых содержит очередной элемент итерируемого объекта iterable, а также предыдущий и следующий за ним элементы:

(<предыдущий элемент>, <очередной элемент>, <следующий элемент>).

Для первого элемента предыдущим считается значение None, для последнего элемента следующим считается так же значение None.

```python
def around(iterable):
    """
    Генераторная функция, возвращающая последовательность кортежей:
    (<предыдущий элемент>, <очередной элемент>, <следующий элемент>).
    """
    it = iter(iterable)  # Создаем итератор из iterable
    previous = None  # Предыдущий элемент (изначально None)
    current = next(it, None)  # Первый элемент (или None, если iterable пуст)

    if current is None:  # Если iterable пуст, завершаем выполнение
        return

    for next_item in it:  # Проходим по оставшимся элементам
        yield (previous, current, next_item)  # Возвращаем кортеж
        previous = current  # Обновляем предыдущий элемент
        current = next_item  # Обновляем текущий элемент

    # Для последнего элемента следующий элемент равен None
    yield (previous, current, None)


# Пример использования
if __name__ == "__main__":
    # Создаем генератор
    generator = around([1, 2, 3, 4])

    # Получаем значения из генератора
    print("Кортежи (предыдущий, текущий, следующий):")
    for value in generator:
        print(value)
```
