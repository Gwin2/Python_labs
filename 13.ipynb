Дан список целых чисел numbers. Удалите из списка все числа-палиндромы и напечатайте результат на одной строке через пробел. Используйте анонимную функцию.
numbers = [18, 191, 9009, 5665, 78, 77, 45, 23, 19991, 908, 8976, 6565, 5665, 10, 1000, 908, 909, 232, 45654, 786]


```python
# Исходный список чисел
numbers = [18, 191, 9009, 5665, 78, 77, 45, 23, 19991, 908, 8976, 6565, 5665, 10, 1000, 908, 909, 232, 45654, 786]

# Фильтрация списка: оставляем только те числа, которые НЕ являются палиндромами
result = list(filter(lambda x: str(x) != str(x)[::-1], numbers))

# Вывод результата
print(*result)
```
