Назовем диапазоном запись двух натуральных чисел через дефис a-b, где a — левая граница диапазона, b — правая граница диапазона, причем a <= b. Диапазон содержит в себе все числа от a до b включительно. Например, диапазон 1-4 содержит числа 1, 2, 3 и 4.

Реализуйте генераторную функцию parse_ranges(), которая принимает один аргумент:

ranges — строка, в которой через запятую указаны диапазоны чисел.

Функция должна возвращать генератор, порождающий последовательность чисел, содержащихся в диапазонах ranges, в порядке возрастания.

```python
def parse_ranges(ranges):
    """
    Генераторная функция, возвращающая последовательность чисел из диапазонов,
    указанных в строке ranges.
    """
    for part in ranges.split(","):  # Разделяем строку по запятым
        a, b = map(int, part.split("-"))  # Разделяем диапазон на левую и правую границы
        yield from range(a, b + 1)  # Порождаем числа из диапазона [a, b]


# Пример использования
if __name__ == "__main__":
    # Строка с диапазонами
    ranges = "1-4,6-8,10-12,15-15,1-3"

    # Создаем генератор
    generator = parse_ranges(ranges)

    # Получаем значения из генератора
    print("Числа из диапазонов:")
    for value in generator:
        print(value, end=" ")
```
