Реализуйте класс Scales, описывающий весы с двумя чашами. При создании экземпляра класс не должен принимать никаких аргументов.
Класс Scales должен иметь три метода экземпляра:
add_right() — метод, принимающий в качестве аргумента массу груза в килограммах и добавляющий на правую чашу весов этот груз
add_left() — метод, принимающий в качестве аргумента массу груза в килограммах и добавляющий на левую чашу весов этот груз
get_result() — метод, возвращающий строку “Весы в равновесии”, если массы грузов на чашах совпадают, “Правая чаша тяжелее” — если правая чаша тяжелее, “Левая чаша тяжелее” — если левая чаша тяжелее


```python
# Определение класса Scales
class Scales:
    # Конструктор класса, инициализирует массы на чашах весов
    def __init__(self):
        self.left_weight = 0  # Масса на левой чаше
        self.right_weight = 0  # Масса на правой чаше

    # Метод add_right() для добавления массы на правую чашу
    def add_right(self, mass):
        if isinstance(mass, (int, float)) and mass >= 0:  # Проверка, что масса — неотрицательное число
            self.right_weight += mass
            print(f"На правую чашу добавлен груз массой {mass} кг. Текущая масса: {self.right_weight} кг.")
        else:
            raise ValueError("Масса должна быть неотрицательным числом.")

    # Метод add_left() для добавления массы на левую чашу
    def add_left(self, mass):
        if isinstance(mass, (int, float)) and mass >= 0:  # Проверка, что масса — неотрицательное число
            self.left_weight += mass
            print(f"На левую чашу добавлен груз массой {mass} кг. Текущая масса: {self.left_weight} кг.")
        else:
            raise ValueError("Масса должна быть неотрицательным числом.")

    # Метод get_result() для определения состояния весов
    def get_result(self):
        if self.left_weight == self.right_weight:
            return "Весы в равновесии"
        elif self.left_weight > self.right_weight:
            return "Левая чаша тяжелее"
        else:
            return "Правая чаша тяжелее"


# Пример использования класса Scales
if __name__ == "__main__":
    # Создание экземпляра класса Scales
    scales = Scales()

    # Добавление массы на чаши
    scales.add_left(5)  # На левую чашу добавлен груз массой 5 кг
    scales.add_right(3)  # На правую чашу добавлен груз массой 3 кг
    scales.add_right(2)  # На правую чашу добавлен груз массой 2 кг

    # Получение результата
    print(scales.get_result())  # Выведет "Весы в равновесии"

    # Добавление дополнительной массы
    scales.add_left(1)  # На левую чашу добавлен груз массой 1 кг
    print(scales.get_result())  # Выведет "Левая чаша тяжелее"

    # Сброс массы (необязательно)
    scales = Scales()  # Создаем новый экземпляр для сброса
```
