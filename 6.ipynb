Реализуйте рекурсивно функцию, которая возвращает n-й член последовательности Фибоначчи.

На множестве целых чисел дана функция f, такая что 
f(n) = n при n < 3 и 
f(n) = f(n-1)f(n-3) + 2f(n-2) при n >= 3. 
Реализуйте f и найдите f(6).


```python
def f(n):
    # Базовые случаи для n < 3
    if n < 3:
        return n
    
    # Рекурсивный случай для n >= 3
    return f(n - 1) * f(n - 3) + 2 * f(n - 2)

# Вычисление f(6)
result = f(6)
print(f"f(6) = {result}")
```
