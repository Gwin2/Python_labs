Напишите функцию arithmetiс(), которая принимает символ одной из пяти арифметических операций (+, -, *, /, **) и возвращает анонимную функцию двух аргументов для соответствующей операции.
Саму arithmetic() удобнее определить через def, а не через анонимные функции.


```python
def arithmetiс(operation):
    """
    Функция, которая принимает символ арифметической операции
    и возвращает анонимную функцию для выполнения этой операции.

    :param operation: Символ операции ('+', '-', '*', '/', '**')
    :return: lambda-функция с двумя аргументами
    """
    if operation == '+':
        return lambda a, b: a + b
    elif operation == '-':
        return lambda a, b: a - b
    elif operation == '*':
        return lambda a, b: a * b
    elif operation == '/':
        return lambda a, b: a / b if b != 0 else 'Ошибка: деление на ноль'
    elif operation == '**':
        return lambda a, b: a ** b
    else:
        raise ValueError("Недопустимая операция. Используйте '+', '-', '*', '/', '**'.")

# Пример использования
if __name__ == "__main__":
    # Создание функций для разных операций
    add = arithmetiс('+')
    subtract = arithmetiс('-')
    multiply = arithmetiс('*')
    divide = arithmetiс('/')
    power = arithmetiс('**')

    # Выполнение операций
    print(add(5, 3))       # Вывод: 8
    print(subtract(5, 3))  # Вывод: 2
    print(multiply(5, 3))  # Вывод: 15
    print(divide(5, 3))    # Вывод: 1.6666666666666667
    print(power(5, 3))     # Вывод: 125
```
