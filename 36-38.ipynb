```python
import math

# Базовый класс Shape
class Shape:
    def area(self):
        raise NotImplementedError("Метод area() должен быть реализован в дочернем классе.")

    def perimeter(self):
        raise NotImplementedError("Метод perimeter() должен быть реализован в дочернем классе.")


# Класс Polygon (многоугольник)
class Polygon(Shape):
    def __init__(self, sides):
        self.sides = sides  # Список длин сторон

    def perimeter(self):
        return sum(self.sides)


# Класс Circle (круг)
class Circle(Shape):
    def __init__(self, radius):
        if radius < 0:
            raise ValueError("Радиус не может быть отрицательным")
        self.radius = radius

    def area(self):
        return math.pi * self.radius**2

    def perimeter(self):
        return 2 * math.pi * self.radius


# Класс Triangle (треугольник)
class Triangle(Polygon):
    def __init__(self, a, b, c):
        super().__init__([a, b, c])
        if not self._is_valid_triangle():
            raise ValueError("Стороны не образуют треугольник")

    def _is_valid_triangle(self):
        a, b, c = self.sides
        return a + b > c and a + c > b and b + c > a

    def area(self):
        # Формула Герона
        s = self.perimeter() / 2
        a, b, c = self.sides
        return math.sqrt(s * (s - a) * (s - b) * (s - c))


# Класс Quadrilateral (четырехугольник)
class Quadrilateral(Polygon):
    def __init__(self, a, b, c, d):
        super().__init__([a, b, c, d])


# Класс Parallelogram (параллелограмм)
class Parallelogram(Quadrilateral):
    def __init__(self, base, side, height):
        super().__init__(base, side, base, side)
        self.height = height

    def area(self):
        return self.sides[0] * self.height


# Класс Rectangle (прямоугольник)
class Rectangle(Parallelogram):
    def __init__(self, width, height):
        super().__init__(width, height, height)


# Класс Square (квадрат)
class Square(Rectangle):
    def __init__(self, side):
        super().__init__(side, side)


# Класс IsoscelesTriangle (равнобедренный треугольник)
class IsoscelesTriangle(Triangle):
    def __init__(self, base, side):
        super().__init__(base, side, side)


# Класс EquilateralTriangle (равносторонний треугольник)
class EquilateralTriangle(Triangle):
    def __init__(self, side):
        super().__init__(side, side, side)
```

```python
if __name__ == "__main__":
    # Создание круга
    circle = Circle(5)
    print(f"Площадь круга: {circle.area():.2f}")
    print(f"Периметр круга: {circle.perimeter():.2f}")

    # Создание треугольника
    triangle = Triangle(3, 4, 5)
    print(f"Площадь треугольника: {triangle.area():.2f}")
    print(f"Периметр треугольника: {triangle.perimeter()}")

    # Создание квадрата
    square = Square(4)
    print(f"Площадь квадрата: {square.area()}")
    print(f"Периметр квадрата: {square.perimeter()}")
```

```python
# Базовый класс LivingBeing
class LivingBeing:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def grow(self):
        self.age += 1
        print(f"{self.name} стал старше на 1 год. Теперь возраст: {self.age}")


# Класс Animal (животное)
class Animal(LivingBeing):
    def __init__(self, name, age, species):
        super().__init__(name, age)
        self.species = species

    def make_sound(self):
        raise NotImplementedError("Метод make_sound() должен быть реализован в дочернем классе.")


# Класс Mammal (млекопитающее)
class Mammal(Animal):
    def __init__(self, name, age, species, fur_color):
        super().__init__(name, age, species)
        self.fur_color = fur_color

    def make_sound(self):
        return "Звуки млекопитающего"


# Класс Bird (птица)
class Bird(Animal):
    def __init__(self, name, age, species, wingspan):
        super().__init__(name, age, species)
        self.wingspan = wingspan

    def make_sound(self):
        return "Чирик-чирик!"


# Класс Plant (растение)
class Plant(LivingBeing):
    def __init__(self, name, age, type):
        super().__init__(name, age)
        self.type = type

    def photosynthesize(self):
        print(f"{self.name} проводит фотосинтез.")


# Класс Tree (дерево)
class Tree(Plant):
    def __init__(self, name, age, type, height):
        super().__init__(name, age, type)
        self.height = height

    def grow(self):
        super().grow()
        self.height += 0.5
        print(f"{self.name} вырос на 0.5 метра. Высота: {self.height:.2f} м")


# Класс Flower (цветок)
class Flower(Plant):
    def __init__(self, name, age, type, color):
        super().__init__(name, age, type)
        self.color = color

    def bloom(self):
        print(f"{self.name} расцвел цветом {self.color}.")
```

```python
if __name__ == "__main__":
    # Создание животного
    mammal = Mammal("Тигр", 5, "Хищник", "Оранжевый")
    print(mammal.make_sound())
    mammal.grow()

    # Создание птицы
    bird = Bird("Воробей", 2, "Певчая птица", 20)
    print(bird.make_sound())

    # Создание дерева
    tree = Tree("Дуб", 50, "Лиственное", 10)
    tree.photosynthesize()
    tree.grow()

    # Создание цветка
    flower = Flower("Роза", 1, "Цветущее", "Красный")
    flower.bloom()
```
