Дан список numbers, состоящий из кортежей. Напишите программу, чтобы список numbers был упорядочен по убыванию среднего арифметического элементов кортежей списка numbers.
numbers = [(10, -2, 3, 4), (-13, 56), (1, 9, 2), (-1, -9, -45, 32), (-1, 5, 1), (17, 0, 1), (0, 1), (3,), (39, 12), (11, -23), (10, -100, 21, 32), (3, -8), (1, 1)]


```python
# Исходный список кортежей
numbers = [(10, -2, 3, 4), (-13, 56), (1, 9, 2), (-1, -9, -45, 32), (-1, 5, 1), 
           (17, 0, 1), (0, 1), (3,), (39, 12), (11, -23), (10, -100, 21, 32), 
           (3, -8), (1, 1)]

# Сортируем список по убыванию среднего арифметического элементов кортежей
sorted_numbers = sorted(numbers, key=lambda x: sum(x) / len(x), reverse=True)

# Выводим результат
print("Отсортированный список:")
for i, tup in enumerate(sorted_numbers, 1):
    avg = sum(tup) / len(tup)
    print(f"{i}. {tup} -> Среднее арифметическое: {avg:.2f}")
```
