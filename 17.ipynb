У одного сервиса знакомств есть списки своих клиентов, парней отдельно и девушек отдельно. Чтобы не заморачиваться с поиском партнёров своим клиентам, работники сервиса предложили суперидею: нужно просто отсортировать два списка клиентов, мужской и женский, в алфавитном порядке их имён, а затем сформировать пары по одинаковым местам в новых списках (например, если первые места заняли Артур и Анна, то они должны стать парой). Помогите работникам сервиса соединить одинокие сердца!❤️
boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard', 'Neal', 'Lucas', 'Diego', 'Juan', 'Zeus', 'Anton', 'Hector', 'Timur', 'Tristan', 'Fushi', 'Vlad', 'Paul', 'Ringo', 'Sam', 'Leo', 'Cigurd', 'Werth', 'Thomas', 'Herbert', 'Kirill', 'Buratino']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha', 'Mary', 'Dora', 'Laura', 'Laurinda', 'Yvete', 'Mayu', 'Sara', 'Gertruda', 'George', 'Polly', 'Irenne', 'Andromache', 'Kumiko', 'Rose', 'Maysa', 'Nazira', 'Jade', 'Latiffa', 'Fatima', 'Edna', 'Malvina' ]


```python
# Исходные списки
boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard', 'Neal', 'Lucas', 'Diego', 'Juan', 'Zeus', 'Anton', 'Hector', 'Timur', 'Tristan', 'Fushi', 'Vlad', 'Paul', 'Ringo', 'Sam', 'Leo', 'Cigurd', 'Werth', 'Thomas', 'Herbert', 'Kirill', 'Buratino']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha', 'Mary', 'Dora', 'Laura', 'Laurinda', 'Yvete', 'Mayu', 'Sara', 'Gertruda', 'George', 'Polly', 'Irenne', 'Andromache', 'Kumiko', 'Rose', 'Maysa', 'Nazira', 'Jade', 'Latiffa', 'Fatima', 'Edna', 'Malvina']

# Шаг 1: Сортируем оба списка в алфавитном порядке
sorted_boys = sorted(boys)
sorted_girls = sorted(girls)

# Шаг 2: Формируем пары
pairs = list(zip(sorted_boys, sorted_girls))

# Шаг 3: Выводим результат
print("Список пар:")
for boy, girl in pairs:
    print(f"{boy} и {girl}")
```
