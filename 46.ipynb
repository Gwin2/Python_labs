Вам дан список чисел. Напишите код, который выдаст 4-й элемент итератора на основе этого списка: numbers = [100, 70, 34, 45, 30, 83, 12, 83, -28, 49, -8, -2, 6, 62, 64, -22, -19, 61, 13, 5, 80, -17, 7, 3, 21, 73, 88, -11, 16, -22]

```python
# Исходный список чисел
numbers = [100, 70, 34, 45, 30, 83, 12, 83, -28, 49, -8, -2, 6, 62, 64, -22, -19, 61, 13, 5, 80, -17, 7, 3, 21, 73, 88, -11, 16, -22]

# Создаем итератор на основе списка
iterator = iter(numbers)

# Получаем 4-й элемент итератора
try:
    # Пропускаем первые 3 элемента
    next(iterator)  # 1-й элемент
    next(iterator)  # 2-й элемент
    next(iterator)  # 3-й элемент
    fourth_element = next(iterator)  # 4-й элемент
    print("4-й элемент итератора:", fourth_element)
except StopIteration:
    print("Список содержит меньше 4 элементов.")
```
