В цветовой схеме RGB цвета задаются с помощью трех компонентов:
R — интенсивность красной составляющей цвета;
G — зеленой;
B — синей.
Противоположные цвета задаются как RGB и (255-R)(255-G)(255-B). Считается, что такие цвета хорошо гармонируют друг с другом.
Напишите программу, которая по трем компонентам заданного цвета, находит компоненты противоположного цвета. 
На вход подаётся строка с числами через пробел и то же самое на выходе, например
122 45 90


```python
# Ввод компонентов цвета как строку через пробел
rgb_input = input("Введите компоненты цвета (R G B): ")

# Преобразование строки в список целых чисел
r, g, b = map(int, rgb_input.split())

# Вычисление противоположных компонентов
opposite_r = 255 - r
opposite_g = 255 - g
opposite_b = 255 - b

# Вывод результатов как строку через пробел
print(f"{opposite_r} {opposite_g} {opposite_b}")
```
