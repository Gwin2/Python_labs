Создайте итератор, который бесконечно возвращает только одну строку: "I love you!"

```python
class InfiniteLove:
    def __iter__(self):
        return self  # Возвращаем сам объект как итератор

    def __next__(self):
        return "I love you!"  # Всегда возвращаем одну и ту же строку


# Пример использования
if __name__ == "__main__":
    love_iterator = InfiniteLove()

    # Выводим первые 5 значений из итератора
    for _ in range(5):
        print(next(love_iterator))
```
