Попробуйте создать свой собственный класс с некоторыми атрибутами. Под него напишите два-три экземпляра. Пусть они будут различаться между собой какими-то своими собственными атрибутами. Убедитесь, что все атрибуты класса принадлежат всем созданными экземплярам, а уникальные атрибуты одного экземпляра не принадлежат другому экземпляру.


```python
# Определение класса Person
class Person:
    # Общие атрибуты класса (принадлежат всем экземплярам)
    species = "Homo sapiens"
    can_speak = True

    # Конструктор класса с уникальными атрибутами для каждого экземпляра
    def __init__(self, name, age, hobby):
        self.name = name  # Уникальный атрибут: имя
        self.age = age    # Уникальный атрибут: возраст
        self.hobby = hobby  # Уникальный атрибут: хобби

    # Метод для вывода информации об экземпляре
    def introduce(self):
        return f"Привет! Меня зовут {self.name}, мне {self.age} лет, и я люблю {self.hobby}."


# Создание экземпляров класса Person
person1 = Person("Александр", 30, "читать книги")
person2 = Person("Елена", 25, "играть на гитаре")

# Проверка общих атрибутов класса
print(f"{person1.name} принадлежит к виду {person1.species}. Может говорить: {person1.can_speak}")
print(f"{person2.name} принадлежит к виду {person2.species}. Может говорить: {person2.can_speak}")

# Проверка уникальных атрибутов каждого экземпляра
print(person1.introduce())
print(person2.introduce())

# Попытка доступа к уникальному атрибуту одного экземпляра через другой экземпляр
try:
    print(person2.name)  # Это работает, так как у person2 есть атрибут name
    print(person2.age)   # Это работает, так как у person2 есть атрибут age
    print(person1.hobby)  # Это работает, так как у person1 есть атрибут hobby
    print(person2.hobby)  # Это работает, так как у person2 есть атрибут hobby
    print(person1.species)  # Это работает, так как species — общий атрибут класса
    print(person2.species)  # Это работает, так как species — общий атрибут класса

    # Попытка получить атрибут другого экземпляра напрямую
    print(person1.name in person2.__dict__)  # Проверяем, является ли name из person1 частью person2
except AttributeError as e:
    print(f"Ошибка: {e}")
```
