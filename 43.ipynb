 Реализуйте класс Vector, описывающий вектор на плоскости. При создании экземпляра класс должен принимать два аргумента в следующем порядке:
x — координата вектора по оси x, по умолчанию имеет значение 0;
y — координата вектора по оси y, по умолчанию имеет значение 0.
Экземпляр класса Vector должен иметь два атрибута:
x — координата вектора по оси x;
y — координата вектора по оси y.
Класс Vector должен иметь два метода экземпляра:
abs() — метод, возвращающий модуль вектора;
orient() — метод, возвращающий направление вектора (в градусах).


```python
import math

# Определение класса Vector
class Vector:
    # Конструктор класса, принимающий координаты x и y
    def __init__(self, x=0, y=0):
        self.x = x  # Координата по оси x
        self.y = y  # Координата по оси y

    # Метод abs() для вычисления модуля вектора
    def abs(self):
        return math.sqrt(self.x**2 + self.y**2)

    # Метод orient() для вычисления направления вектора в градусах
    def orient(self):
        # Вычисляем угол в радианах с помощью atan2
        angle_radians = math.atan2(self.y, self.x)
        # Преобразуем угол в градусы
        angle_degrees = math.degrees(angle_radians)
        # Если угол отрицательный, переводим его в диапазон [0, 360)
        if angle_degrees < 0:
            angle_degrees += 360
        return angle_degrees


# Пример использования класса Vector
if __name__ == "__main__":
    # Создание экземпляра класса Vector
    vector1 = Vector(3, 4)  # Вектор (3, 4)
    vector2 = Vector(1, -1)  # Вектор (1, -1)

    # Вычисление модуля вектора
    print(f"Модуль вектора {vector1.x, vector1.y}: {vector1.abs():.2f}")
    print(f"Модуль вектора {vector2.x, vector2.y}: {vector2.abs():.2f}")

    # Вычисление направления вектора
    print(f"Направление вектора {vector1.x, vector1.y}: {vector1.orient():.2f} градусов")
    print(f"Направление вектора {vector2.x, vector2.y}: {vector2.orient():.2f} градусов")
```
