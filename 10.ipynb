Дан некоторый диапазон целых чисел и несколько числовых отрезков в пределах этого диапазона.  Найти суммарную длину, перекрываемую отрезками. Не забывайте, что отрезки могут пересекаться и не один раз.

```python
def total_covered_length(segments):
    """
    Функция для вычисления суммарной длины, перекрываемой отрезками.
    
    :param segments: Список отрезков в формате [(start1, end1), (start2, end2), ...]
    :return: Общая длина, перекрываемая отрезками
    """
    if not segments:
        return 0

    # Шаг 1: Сортировка отрезков по начальной точке
    segments.sort(key=lambda x: x[0])

    # Шаг 2: Объединение пересекающихся отрезков
    merged_segments = [segments[0]]  # Начинаем с первого отрезка
    for current_start, current_end in segments[1:]:
        last_start, last_end = merged_segments[-1]

        # Если текущий отрезок пересекается с последним объединенным отрезком
        if current_start <= last_end:
            # Объединяем их, расширяя последний отрезок
            merged_segments[-1] = (last_start, max(last_end, current_end))
        else:
            # Иначе добавляем текущий отрезок как новый
            merged_segments.append((current_start, current_end))

    # Шаг 3: Вычисление общей длины
    total_length = sum(end - start for start, end in merged_segments)
    return total_length


# Пример использования
if __name__ == "__main__":
    # Список отрезков
    segments = [(1, 4), (3, 5), (7, 10), (9, 12), (15, 16)]

    # Вычисление общей длины
    result = total_covered_length(segments)
    print(f"Суммарная длина, перекрываемая отрезками: {result}")
```
