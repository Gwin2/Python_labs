Реализуйте класс Gun, описывающий ружье. При создании экземпляра класс не должен принимать никаких аргументов.
Класс Gun должен иметь три метода экземпляра:
shoot() — метод, при первом вызове которого выводится строка pif, при втором — paf, при третьем — pif, при четвертом — paf, и так далее;
shots_count() — метод, возвращающий актуальное количество вызовов метода shoot();
shots_reset() — метод, сбрасывающий количество вызовов метода shoot() до нуля.


```python
# Определение класса Gun
class Gun:
    # Конструктор класса, инициализирует счетчик выстрелов
    def __init__(self):
        self.shot_count = 0  # Счетчик вызовов метода shoot()

    # Метод shoot() для стрельбы
    def shoot(self):
        if self.shot_count % 2 == 0:  # Если четное количество выстрелов (0, 2, 4, ...)
            print("pif")
        else:  # Если нечетное количество выстрелов (1, 3, 5, ...)
            print("paf")
        self.shot_count += 1  # Увеличиваем счетчик выстрелов

    # Метод shots_count() для получения количества выстрелов
    def shots_count(self):
        return self.shot_count

    # Метод shots_reset() для сброса счетчика выстрелов
    def shots_reset(self):
        self.shot_count = 0
        print("Счетчик выстрелов сброшен до нуля.")


# Пример использования класса Gun
if __name__ == "__main__":
    # Создание экземпляра класса Gun
    my_gun = Gun()

    # Вызов метода shoot()
    my_gun.shoot()  # Выведет "pif"
    my_gun.shoot()  # Выведет "paf"
    my_gun.shoot()  # Выведет "pif"
    my_gun.shoot()  # Выведет "paf"

    # Получение количества выстрелов
    print(f"Количество выстрелов: {my_gun.shots_count()}")  # Выведет "4"

    # Сброс счетчика выстрелов
    my_gun.shots_reset()

    # Проверка сброса
    print(f"Количество выстрелов после сброса: {my_gun.shots_count()}")  # Выведет "0"
```
