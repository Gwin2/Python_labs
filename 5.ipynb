Напишите функцию, которая для заданных в аргументах объектов, возвращает как результат их перевернутый список.


```python
def reverse_objects(*args):
    """
    Функция принимает произвольное количество аргументов и возвращает их в виде перевернутого списка.
    """
    # Создаем список из переданных аргументов и переворачиваем его
    return list(args)[::-1]

# Тестирование функции
if __name__ == "__main__":
    result = reverse_objects(10, "hello", [1, 2, 3], {"key": "value"}, 42)
    print("Перевернутый список:", result)
```
