1) PAN (Permanent Account Number) – это уникальный номер, который присваивается всем налогоплательщикам в Индии. Он имеет следующий формат:
<letter><letter><letter><letter><letter><digit><digit><digit><digit><letter>
где letter — заглавная латинская буква, а digit — цифра.
Какому регулярному выражению соответствует PAN?
2) Какому регулярному выражению соответствует формат времени ЧЧ:ММ?
3) Попробуйте теперь решить задачу на телефонные номера. Учтите, что кто-то пишет номера с 8 или +7 в начале, кто-то пишет через пробелы или без разделителей (сплошные цифры).

```python
import re

# Регулярное выражение для PAN
pan_pattern = r'^[A-Z]{5}\d{4}[A-Z]$'

# Функция для проверки PAN
def is_valid_pan(pan):
    return bool(re.match(pan_pattern, pan))

# Примеры использования
test_pans = ["ABCDE1234F", "abcde1234F", "ABCDE12345", "ABCDE123FG"]
for pan in test_pans:
    print(f"{pan}: {'Valid' if is_valid_pan(pan) else 'Invalid'}")
```

```python
# Регулярное выражение для времени
time_pattern = r'^(?:[01]\d|2[0-3]):[0-5]\d$'

# Функция для проверки времени
def is_valid_time(time_str):
    return bool(re.match(time_pattern, time_str))

# Примеры использования
test_times = ["12:34", "23:59", "00:00", "24:00", "12:60", "1:23"]
for time_str in test_times:
    print(f"{time_str}: {'Valid' if is_valid_time(time_str) else 'Invalid'}")
```

```python
# Регулярное выражение для телефонных номеров
phone_pattern = r'^(?:\+7|8)[-\s]?(\d{3})[-\s]?(\d{3})[-\s]?(\d{2})[-\s]?(\d{2})$'

# Функция для проверки телефонного номера
def is_valid_phone(phone):
    return bool(re.match(phone_pattern, phone))

# Примеры использования
test_phones = [
    "89123456789",
    "+79123456789",
    "8 912 345 67 89",
    "+7 912 345 6789",
    "8-912-345-67-89",
    "+7-912-345-6789",
    "8912345",
    "+791234567890",
    "abc123"
]

for phone in test_phones:
    print(f"{phone}: {'Valid' if is_valid_phone(phone) else 'Invalid'}")
```
