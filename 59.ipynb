Реализуйте генераторную функцию palindromes(), которая не принимает никаких аргументов.

Функция должна возвращать генератор, порождающий бесконечную последовательность натуральных чисел-палиндромов.

```python
def is_palindrome(n):
    """Вспомогательная функция для проверки, является ли число палиндромом."""
    return str(n) == str(n)[::-1]  # Проверяем, совпадает ли число с его перевёрнутой версией

def palindromes():
    """Генераторная функция для порождения бесконечной последовательности палиндромов."""
    n = 1  # Начинаем с первого натурального числа
    while True:  # Бесконечный цикл
        if is_palindrome(n):  # Проверяем, является ли число палиндромом
            yield n  # Возвращаем число, если оно палиндром
        n += 1  # Переходим к следующему натуральному числу


# Пример использования
if __name__ == "__main__":
    # Создаем генератор
    generator = palindromes()

    # Получаем первые 10 палиндромов
    print("Первые 10 палиндромов:")
    for _ in range(10):
        print(next(generator), end=" ")
```
