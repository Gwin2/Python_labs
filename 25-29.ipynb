1) Напишите программу, которая на входе принимает строку с именем текстового файла и выводит на экран содержимое этого файла.
2) Напишите программу, которая на входе принимает строку с именем текстового файла и выводит на экран предпоследнюю строку из содержимого этого файла.
3) Напишите программу, которая на входе принимает строку с именем текстового файла и выводит на экран случайную строку из содержимого этого файла.
4) Напишите программу, которая работает с текстовым файлом, содержащим только числа, отделённые друг от друга пробелом или переносом строки. На выходе программа выдаёт сумму всех чисел, содержащихся в файле.
5) Напишите программу, принимающую на вход текстовый файл (текст на латинице), которая выводит на отдельных строках количества строк, слов и букв в файле. 


```python
# Открываем файл и выводим его содержимое
def print_file_content(filename):
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            print(file.read())
    except FileNotFoundError:
        print("Файл не найден.")
    except Exception as e:
        print(f"Произошла ошибка: {e}")

# Пример использования
if __name__ == "__main__":
    filename = input("Введите имя файла: ")
    print_file_content(filename)
```

```python
# Находим предпоследнюю строку в файле
def print_second_last_line(filename):
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            lines = file.readlines()  # Читаем все строки
            if len(lines) < 2:
                print("В файле недостаточно строк для вывода предпоследней.")
            else:
                print(lines[-2].strip())  # Выводим предпоследнюю строку
    except FileNotFoundError:
        print("Файл не найден.")
    except Exception as e:
        print(f"Произошла ошибка: {e}")

# Пример использования
if __name__ == "__main__":
    filename = input("Введите имя файла: ")
    print_second_last_line(filename)
```

```python
import random

# Вывод случайной строки из файла
def print_random_line(filename):
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            lines = file.readlines()  # Читаем все строки
            if not lines:
                print("Файл пустой.")
            else:
                print(random.choice(lines).strip())  # Выбираем случайную строку
    except FileNotFoundError:
        print("Файл не найден.")
    except Exception as e:
        print(f"Произошла ошибка: {e}")

# Пример использования
if __name__ == "__main__":
    filename = input("Введите имя файла: ")
    print_random_line(filename)
```

```python
# Вычисление суммы чисел в файле
def sum_numbers_in_file(filename):
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            content = file.read().replace('\n', ' ')  # Заменяем переносы строки на пробелы
            numbers = list(map(int, content.split()))  # Преобразуем строки в числа
            print(f"Сумма чисел в файле: {sum(numbers)}")
    except FileNotFoundError:
        print("Файл не найден.")
    except ValueError:
        print("Ошибка: файл содержит нечисловые данные.")
    except Exception as e:
        print(f"Произошла ошибка: {e}")

# Пример использования
if __name__ == "__main__":
    filename = input("Введите имя файла: ")
    sum_numbers_in_file(filename)
```

```python
# Подсчет строк, слов и букв в файле
def count_lines_words_chars(filename):
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            lines = file.readlines()  # Читаем все строки
            num_lines = len(lines)  # Количество строк
            num_words = sum(len(line.split()) for line in lines)  # Количество слов
            num_chars = sum(len(line.replace(' ', '').replace('\n', '')) for line in lines)  # Количество букв
            
            print(f"Количество строк: {num_lines}")
            print(f"Количество слов: {num_words}")
            print(f"Количество букв: {num_chars}")
    except FileNotFoundError:
        print("Файл не найден.")
    except Exception as e:
        print(f"Произошла ошибка: {e}")

# Пример использования
if __name__ == "__main__":
    filename = input("Введите имя файла: ")
    count_lines_words_chars(filename)
```
