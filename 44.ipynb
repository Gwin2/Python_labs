 Будем называть словом любую последовательность из одной или более букв. Текстом будем считать набор слов, разделенных пробельными символами.
Реализуйте класс Text, описывающий изначально пустой расширяемый набор слов. При создании экземпляра класс не должен принимать никаких аргументов.
Экземпляр класса Text должен иметь три метода:
add_words() — метод, принимающий в качестве аргумента текст и добавляющий слова из данного текста в набор;
get_shortest_words() — метод, возвращающий актуальный список самых коротких слов в наборе;
get_longest_words() — метод, возвращающий актуальный список самых длинных слов в наборе.
Слова в списках, возвращаемых методами get_shortest_words() и get_longest_words(), должны располагаться в том порядке, в котором они были добавлены в набор.


```python
# Определение класса Text
class Text:
    # Конструктор класса, инициализирует пустой список слов
    def __init__(self):
        self.words = []  # Список для хранения всех добавленных слов

    # Метод add_words() для добавления слов из текста
    def add_words(self, text):
        # Разделяем текст на слова по пробельным символам
        new_words = text.split()
        # Добавляем новые слова в общий список
        self.words.extend(new_words)

    # Метод get_shortest_words() для получения списка самых коротких слов
    def get_shortest_words(self):
        if not self.words:  # Если список слов пуст, возвращаем пустой список
            return []
        # Находим минимальную длину слова
        min_length = min(len(word) for word in self.words)
        # Фильтруем слова с минимальной длиной
        return [word for word in self.words if len(word) == min_length]

    # Метод get_longest_words() для получения списка самых длинных слов
    def get_longest_words(self):
        if not self.words:  # Если список слов пуст, возвращаем пустой список
            return []
        # Находим максимальную длину слова
        max_length = max(len(word) for word in self.words)
        # Фильтруем слова с максимальной длиной
        return [word for word in self.words if len(word) == max_length]


# Пример использования класса Text
if __name__ == "__main__":
    # Создание экземпляра класса Text
    text_processor = Text()

    # Добавление слов в набор
    text_processor.add_words("This is a simple test")
    text_processor.add_words("Another example with longer words")

    # Получение самых коротких слов
    print("Самые короткие слова:", text_processor.get_shortest_words())

    # Получение самых длинных слов
    print("Самые длинные слова:", text_processor.get_longest_words())
```
