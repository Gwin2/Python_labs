Реализуйте класс Gun, описывающий ружье. При создании экземпляра класс не должен принимать никаких аргументов.
Класс Gun должен иметь один метод экземпляра:
shoot() — метод, при вызове которого выводится строка pif


```python
# Определение класса Gun
class Gun:
    # Конструктор класса (не принимает аргументов)
    def __init__(self):
        pass  # Ничего не делаем при инициализации

    # Метод shoot() для вывода строки "pif"
    def shoot(self):
        print("pif")


# Пример использования класса Gun
if __name__ == "__main__":
    # Создание экземпляра класса Gun
    my_gun = Gun()

    # Вызов метода shoot()
    my_gun.shoot()  # Выведет: pif
```
