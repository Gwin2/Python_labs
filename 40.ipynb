Реализуйте класс Ant, описывающий муравья, которая перемещается по координатной плоскости в четырех направлениях: вверх, вниз, влево и вправо. При создании экземпляра класс должен принимать два аргумента в следующем порядке: x, y. По умолчанию оба аргумента имеют значение 0.
Класс должен иметь четыре метода экземпляра:
move_up() — метод, принимающий в качестве аргумента целое число n и увеличивающий координату по оси y на n
move_down() — метод, принимающий в качестве аргумента целое число n и уменьшающий координату по оси y на n
move_right() — метод, принимающий в качестве аргумента целое число n и увеличивающий координату по оси x на n
move_left() — метод, принимающий в качестве аргумента целое число n и уменьшающий координату по оси x на n


```python
# Определение класса Ant
class Ant:
    # Конструктор класса, принимающий начальные координаты x и y
    def __init__(self, x=0, y=0):
        self.x = x  # Координата по оси x
        self.y = y  # Координата по оси y

    # Метод move_up() для перемещения вверх
    def move_up(self, n):
        if isinstance(n, int) and n > 0:  # Проверка, что n — целое положительное число
            self.y += n
            print(f"Муравей переместился вверх на {n}. Текущие координаты: ({self.x}, {self.y})")
        else:
            raise ValueError("Перемещение должно быть целым положительным числом.")

    # Метод move_down() для перемещения вниз
    def move_down(self, n):
        if isinstance(n, int) and n > 0:  # Проверка, что n — целое положительное число
            self.y -= n
            print(f"Муравей переместился вниз на {n}. Текущие координаты: ({self.x}, {self.y})")
        else:
            raise ValueError("Перемещение должно быть целым положительным числом.")

    # Метод move_right() для перемещения вправо
    def move_right(self, n):
        if isinstance(n, int) and n > 0:  # Проверка, что n — целое положительное число
            self.x += n
            print(f"Муравей переместился вправо на {n}. Текущие координаты: ({self.x}, {self.y})")
        else:
            raise ValueError("Перемещение должно быть целым положительным числом.")

    # Метод move_left() для перемещения влево
    def move_left(self, n):
        if isinstance(n, int) and n > 0:  # Проверка, что n — целое положительное число
            self.x -= n
            print(f"Муравей переместился влево на {n}. Текущие координаты: ({self.x}, {self.y})")
        else:
            raise ValueError("Перемещение должно быть целым положительным числом.")


# Пример использования класса Ant
if __name__ == "__main__":
    # Создание экземпляра класса Ant с начальными координатами (0, 0)
    ant = Ant()

    # Перемещение муравья
    ant.move_up(3)      # Вверх
    ant.move_right(5)   # Вправо
    ant.move_down(2)    # Вниз
    ant.move_left(4)    # Влево

    # Попытка передать отрицательное значение
    try:
        ant.move_up(-1)
    except ValueError as e:
        print(e)
```
