Реализуйте класс Square, порождающий итераторы, конструктор которого принимает один аргумент:

n — натуральное число.

Итератор класса Square должен генерировать последовательность из n чисел, каждое из которых является квадратом очередного натурального числа, а затем возбуждать исключение StopIteration.

```python
class Square:
    def __init__(self, n):
        self.n = n  # Количество чисел в последовательности
        self.current = 1  # Текущее натуральное число

    def __iter__(self):
        return self  # Возвращаем сам объект как итератор

    def __next__(self):
        if self.current <= self.n:  # Проверяем, не исчерпано ли количество чисел
            square = self.current ** 2  # Вычисляем квадрат текущего числа
            self.current += 1  # Переходим к следующему натуральному числу
            return square  # Возвращаем квадрат
        else:
            raise StopIteration  # Возбуждаем исключение, если достигнуто ограничение


# Пример использования
if __name__ == "__main__":
    # Создаем итератор, который генерирует квадраты первых 5 натуральных чисел
    squares = Square(5)

    # Используем цикл for для перебора значений
    for value in squares:
        print(value)
```
