Реализуйте декоратор reverse_args, который передает все позиционные аргументы в декорируемую функцию func в обратном порядке.

```python
def reverse_args(func):
    """
    Декоратор, который передает все позиционные аргументы
    в декорируемую функцию func в обратном порядке.
    """
    def wrapper(*args, **kwargs):
        # Передаем позиционные аргументы в обратном порядке
        reversed_args = args[::-1]
        return func(*reversed_args, **kwargs)
    
    return wrapper


# Пример использования
if __name__ == "__main__":
    @reverse_args
    def concat(a, b, c):
        return f"{a}{b}{c}"

    # Вызываем декорированную функцию
    result = concat("A", "B", "C")
    print(result)  # Вывод: CBA
```
