Реализуйте класс BoundedRepeater, порождающий итераторы, конструктор которого принимает два аргумента в следующем порядке:

obj — произвольный объект; times — натуральное число.

Итератор класса BoundedRepeater должен генерировать значение obj times раз, а затем возбуждать исключение StopIteration.

```python
class BoundedRepeater:
    def __init__(self, obj, times):
        self.obj = obj  # Сохраняем объект, который будет возвращаться
        self.times = times  # Количество повторений
        self.counter = 0  # Счетчик текущих повторений

    def __iter__(self):
        return self  # Возвращаем сам объект как итератор

    def __next__(self):
        if self.counter < self.times:  # Проверяем, не исчерпано ли количество повторений
            self.counter += 1  # Увеличиваем счетчик
            return self.obj  # Возвращаем объект
        else:
            raise StopIteration  # Возбуждаем исключение, если достигнуто ограничение


# Пример использования
if __name__ == "__main__":
    # Создаем итератор, который возвращает строку "Hello" 3 раза
    repeater = BoundedRepeater("Hello", 3)

    # Используем цикл for для перебора значений
    for value in repeater:
        print(value)
```
