На вход программе подается строка текста. Напишите программу, которая преобразует каждое слово введенного текста в "молодежный жаргон" по следующему правилу:
1) первая буква каждого слова удаляется и ставится в конец слова; 2) последняя буква точно так же удаляется и переносится в начало слова; 3) затем каждая гласная, кроме йотированных, заменяется на "ё".


```python
# Ввод строки текста
text = input("Введите строку текста: ")

# Множество гласных без йотированных
vowels = "aeiouAEIOUаеиоуыэАЕИОУЫЭ"

# Функция для преобразования одного слова
def transform_word(word):
    if len(word) < 2:  # Если слово короче двух букв, возвращаем его без изменений
        return word
    
    # Шаг 1: Перемещаем первую букву в конец слова
    first_char = word[0]
    word = word[1:] + first_char
    
    # Шаг 2: Перемещаем последнюю букву в начало слова
    last_char = word[-1]
    word = last_char + word[:-1]
    
    # Шаг 3: Заменяем гласные на "ё"
    transformed_word = ''.join(['ё' if char in vowels else char for char in word])
    
    return transformed_word

# Преобразуем каждое слово в строке
result = ' '.join([transform_word(word) for word in text.split()])

# Выводим результат
print("Текст в молодежном жаргоне:")
print(result)
```
