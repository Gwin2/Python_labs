Создайте класс House, описывающий дом. При создании экземпляра класс принимает два аргумента в следующем порядке:
color — цвет дома;
rooms — количество комнат в доме.
Экземпляр данного класса имеет два атрибута:
color — цвет дома;
rooms — количество комнат в доме.
Реализуйте для класса House два метода экземпляра:
paint() — метод, принимающий в качестве аргумента значение new_color и изменяющий текущий цвет дома на new_color;
add_rooms() — метод, принимающий в качестве аргумента целое число n и увеличивающий количество комнат в доме на n.


```python
# Определение класса House
class House:
    # Конструктор класса, принимающий цвет дома и количество комнат
    def __init__(self, color, rooms):
        self.color = color  # Цвет дома
        self.rooms = rooms  # Количество комнат в доме

    # Метод paint() для изменения цвета дома
    def paint(self, new_color):
        self.color = new_color
        print(f"Дом теперь окрашен в {self.color} цвет.")

    # Метод add_rooms() для увеличения количества комнат
    def add_rooms(self, n):
        if isinstance(n, int) and n > 0:  # Проверка, что n — целое положительное число
            self.rooms += n
            print(f"Количество комнат увеличено на {n}. Теперь в доме {self.rooms} комнат.")
        else:
            raise ValueError("Количество добавляемых комнат должно быть целым положительным числом.")


# Пример использования класса House
if __name__ == "__main__":
    # Создание экземпляра класса House
    my_house = House("красный", 3)

    # Вывод начальных значений атрибутов
    print(f"Цвет дома: {my_house.color}")
    print(f"Количество комнат: {my_house.rooms}")

    # Изменение цвета дома
    my_house.paint("синий")

    # Добавление комнат
    my_house.add_rooms(2)

    # Попытка добавить отрицательное количество комнат
    try:
        my_house.add_rooms(-1)
    except ValueError as e:
        print(e)
```
