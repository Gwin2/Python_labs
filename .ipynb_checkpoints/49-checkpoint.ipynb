Реализуйте функцию get_min_max(), которая принимает один аргумент:

iterable — итерируемый объект, элементы которого сравнимы между собой.

Функция должна возвращать кортеж, в котором первым элементом является минимальный элемент итерируемого объекта iterable, вторым — максимальный элемент итерируемого объекта iterable. Если итерируемый объект iterable пуст, функция должна вернуть значение None.

```python
def get_min_max(iterable):
    # Преобразуем итерируемый объект в итератор
    iterator = iter(iterable)
    
    try:
        # Получаем первый элемент, чтобы инициализировать min_value и max_value
        first = next(iterator)
        min_value = max_value = first
        
        # Проходим по оставшимся элементам и обновляем min_value и max_value
        for item in iterator:
            if item < min_value:
                min_value = item
            if item > max_value:
                max_value = item
        
        return (min_value, max_value)
    
    except StopIteration:
        # Если итератор пуст, возвращаем None
        return None


# Пример использования
if __name__ == "__main__":
    # Пример 1: Список чисел
    numbers = [10, 2, 15, 7, 3]
    print(get_min_max(numbers))  # Вывод: (2, 15)

    # Пример 2: Пустой список
    empty_list = []
    print(get_min_max(empty_list))  # Вывод: None

    # Пример 3: Строка (символы сравнимы между собой)
    text = "python"
    print(get_min_max(text))  # Вывод: ('h', 'y')

    # Пример 4: Генератор
    generator = (x for x in range(5))
    print(get_min_max(generator))  # Вывод: (0, 4)
```
