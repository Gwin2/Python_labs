 Реализуйте класс Circle, описывающий круг. При создании экземпляра класс должен принимать один аргумент:
radius — радиус круга
Экземпляр класса Circle должен иметь четыре атрибута:
radius — радиус круга
diameter — диаметр круга
area — площадь круга
length — длина окружности


```python
import math

# Определение класса Circle
class Circle:
    # Конструктор класса, принимающий радиус круга
    def __init__(self, radius):
        if radius < 0:
            raise ValueError("Радиус не может быть отрицательным")
        
        self.radius = radius  # Радиус круга
        self.diameter = 2 * radius  # Диаметр круга
        self.area = math.pi * radius ** 2  # Площадь круга
        self.length = 2 * math.pi * radius  # Длина окружности


# Пример использования класса Circle
if __name__ == "__main__":
    # Создание экземпляра класса Circle
    circle = Circle(5)

    # Вывод атрибутов круга
    print(f"Радиус: {circle.radius}")
    print(f"Диаметр: {circle.diameter}")
    print(f"Площадь: {circle.area:.2f}")
    print(f"Длина окружности: {circle.length:.2f}")
```
