На вход программе подается строка текста, содержащая натуральные числа. Напишите программу, которая вставляет между каждым числом знак “+”, после всех ставит “=”, а затем пишет сумму полученных чисел.

```python
# Ввод строки с числами
input_string = input("Введите строку с числами: ")

# Проверка на пустой ввод
if not input_string.strip():
    print("Ошибка: Вы ничего не ввели.")
else:
    # Шаг 1: Разделение строки на список элементов
    elements = input_string.split()

    # Шаг 2: Проверка, являются ли все элементы числами
    if all(element.isdigit() for element in elements):
        # Преобразование списка строк в список целых чисел
        numbers_int = list(map(int, elements))

        # Шаг 3: Формирование строки с "+", "=", и суммой
        result_string = " + ".join(elements) + f" = {sum(numbers_int)}"

        # Шаг 4: Вывод результата
        print(result_string)
    else:
        print("Ошибка: Ввод должен содержать только натуральные числа, разделенные пробелами.")
```
