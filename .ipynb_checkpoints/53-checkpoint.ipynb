Реализуйте класс Repeater, порождающий итераторы, конструктор которого принимает один аргумент:

obj — произвольный объект.

Итератор класса Repeater должен бесконечно генерировать единственное значение — obj.

```python
class Repeater:
    def __init__(self, obj):
        self.obj = obj  # Сохраняем объект, который будет возвращаться

    def __iter__(self):
        return self  # Возвращаем сам объект как итератор

    def __next__(self):
        return self.obj  # Всегда возвращаем одно и то же значение


# Пример использования
if __name__ == "__main__":
    # Создаем итератор, который всегда возвращает строку "Python"
    repeater = Repeater("Python")

    # Выводим первые 5 значений из итератора
    for _ in range(5):
        print(next(repeater))
```
