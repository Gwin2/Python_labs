Реализуйте через map и reduce функцию, которая вычисляет значения некоторого многочлена произвольной степени от целого x.
Сначала через пробел на одной строке вводятся коэффициенты многочлена, начиная со старшего. Затем на следующей строке число x.
Пример:
2 4 3
10
2 * 10^2 + 4 * 10 + 3 = 243


```python
from functools import reduce

# Ввод коэффициентов многочлена
coefficients = list(map(int, input("Введите коэффициенты многочлена через пробел: ").split()))

# Ввод значения x
x = int(input("Введите значение x: "))

# Вычисление значения многочлена через reduce и метод Горнера
result = reduce(lambda acc, coeff: acc * x + coeff, coefficients, 0)

# Вывод результата
print(f"Значение многочлена при x = {x}: {result}")
```
