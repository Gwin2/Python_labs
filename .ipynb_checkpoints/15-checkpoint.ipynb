Напишите программу, которая с помощью встроенных функций filter() и sorted() выводит слова из списка words, имеющие длину от 5 до 10 включительно символов. Слова следует вывести в алфавитном порядке на одной строке через пробел.
words = ["напиток", "понедельник", "заграница", "предвзятость", "злоупотребление", "отмена", "предложение", "среда", "способный", "предать", "случай", "похитить", "фанатик", "ставка", "отказаться", "кроме", "доступ", "пятница", "отдавать", "изобиловать", "отсутствует", "остерегаться", "изобилие", "ненормальный", "борт", "около", "ускорить", "прервать", 'четверг', 'вторник', 'воскресенье', 'причал', 'предел', 'благожелательный', 'утихать', 'пребывать', 'велосипед', 'рядом', 'принять', 'ягода', 'недоумение', 'резкий', 'суббота', "аксессуар", "поглощать"]


```python
# Исходный список слов
words = ["напиток", "понедельник", "заграница", "предвзятость", "злоупотребление", "отмена", "предложение", "среда", "способный", "предать", "случай", "похитить", "фанатик", "ставка", "отказаться", "кроме", "доступ", "пятница", "отдавать", "изобиловать", "отсутствует", "остерегаться", "изобилие", "ненормальный", "борт", "около", "ускорить", "прервать", 'четверг', 'вторник', 'воскресенье', 'причал', 'предел', 'благожелательный', 'утихать', 'пребывать', 'велосипед', 'рядом', 'принять', 'ягода', 'недоумение', 'резкий', 'суббота', "аксессуар", "поглощать"]

# Отфильтровываем слова с длиной от 5 до 10 символов включительно
filtered_words = filter(lambda word: 5 <= len(word) <= 10, words)

# Сортируем результат в алфавитном порядке
sorted_words = sorted(filtered_words)

# Выводим результат на одной строке через пробел
print(*sorted_words)
```
