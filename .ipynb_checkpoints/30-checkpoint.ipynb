Напишите программу, которая принимает n строк и выводит сумму чисел, чьё строковое представление есть среди введённых строк, а затем выводит в алфавитном порядке все оставшиеся строки, которые не представляют собой числа.

```python
def is_number(s):
    """
    Проверяет, является ли строка числом (целым или с плавающей точкой).
    """
    try:
        float(s)  # Пытаемся преобразовать строку в число
        return True
    except ValueError:
        return False


def process_strings(n):
    total_sum = 0.0  # Переменная для суммы чисел
    non_numeric_strings = []  # Список для хранения строк, которые не являются числами

    print("Введите строки:")
    for _ in range(n):
        s = input().strip()  # Считываем строку и убираем лишние пробелы
        if is_number(s):  # Проверяем, является ли строка числом
            total_sum += float(s)  # Прибавляем число к общей сумме
        else:
            non_numeric_strings.append(s)  # Добавляем строку в список нечисловых строк

    # Выводим сумму чисел
    print(f"Сумма чисел: {total_sum}")

    # Выводим отсортированные нечисловые строки
    sorted_non_numeric = sorted(non_numeric_strings)  # Сортируем строки в алфавитном порядке
    print("Оставшиеся строки (в алфавитном порядке):")
    print(" ".join(sorted_non_numeric))


# Основная часть программы
if __name__ == "__main__":
    try:
        n = int(input("Введите количество строк (n): "))  # Ввод количества строк
        if n <= 0:
            print("Количество строк должно быть положительным.")
        else:
            process_strings(n)
    except ValueError:
        print("Ошибка: введите целое положительное число для n.")
```
