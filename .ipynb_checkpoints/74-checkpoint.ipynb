Реализуйте декоратор takes_positive, который проверяет, что все аргументы, передаваемые в декорируемую функцию, являются положительными целыми числами.

Если хотя бы один аргумент не удовлетворяет данному условию, декоратор должен возбуждать исключение:

TypeError, если аргумент не является целым числом

ValueError, если аргумент является целым числом, но отрицательным или равным нулю

```python
def takes_positive(func):
    """
    Декоратор, который проверяет, что все аргументы функции
    являются положительными целыми числами.
    """
    def wrapper(*args, **kwargs):
        # Проверяем позиционные аргументы
        for arg in args:
            if not isinstance(arg, int):  # Если аргумент не является целым числом
                raise TypeError("Аргумент должен быть целым числом")
            if arg <= 0:  # Если аргумент отрицательный или равен нулю
                raise ValueError("Аргумент должен быть положительным целым числом")
        
        # Проверяем именованные аргументы
        for key, value in kwargs.items():
            if not isinstance(value, int):  # Если аргумент не является целым числом
                raise TypeError(f"Аргумент '{key}' должен быть целым числом")
            if value <= 0:  # Если аргумент отрицательный или равен нулю
                raise ValueError(f"Аргумент '{key}' должен быть положительным целым числом")
        
        # Если все проверки пройдены, вызываем декорируемую функцию
        return func(*args, **kwargs)
    
    return wrapper


# Пример использования
if __name__ == "__main__":
    @takes_positive
    def add(a, b):
        return a + b

    @takes_positive
    def multiply(*args):
        result = 1
        for num in args:
            result *= num
        return result

    # Вызываем декорированные функции
    try:
        print(add(2, 3))  # Вывод: 5
        print(add(-1, 3))  # Возбуждает ValueError
    except (TypeError, ValueError) as e:
        print(e)

    try:
        print(multiply(2, 4, 6))  # Вывод: 48
        print(multiply(2, -4, 6))  # Возбуждает ValueError
    except (TypeError, ValueError) as e:
        print(e)
```
