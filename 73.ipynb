Реализуйте декоратор exception_decorator, который возвращает:

кортеж (value, 'Функция выполнилась без ошибок'), если декорируемая функция завершила свою работу без ошибок, где value — возвращаемое значение декорируемой функции;

кортеж (None, 'При вызове функции произошла ошибка'), если при выполнении декорируемой функции возникла ошибка.

```python
def exception_decorator(func):
    """
    Декоратор, который возвращает кортеж:
    - (value, 'Функция выполнилась без ошибок'), если функция завершилась без ошибок;
    - (None, 'При вызове функции произошла ошибка'), если возникла ошибка.
    """
    def wrapper(*args, **kwargs):
        try:
            # Пытаемся выполнить функцию
            result = func(*args, **kwargs)
            return result, "Функция выполнилась без ошибок"
        except Exception:
            # Если возникла ошибка, возвращаем None и сообщение об ошибке
            return None, "При вызове функции произошла ошибка"
    
    return wrapper


# Пример использования
if __name__ == "__main__":
    @exception_decorator
    def divide(a, b):
        return a / b

    @exception_decorator
    def greet(name):
        return f"Hello, {name}!"

    # Вызываем декорированные функции
    print(divide(10, 2))  # Вывод: (5.0, 'Функция выполнилась без ошибок')
    print(divide(10, 0))  # Вывод: (None, 'При вызове функции произошла ошибка')
    print(greet("Alice"))  # Вывод: ('Hello, Alice!', 'Функция выполнилась без ошибок')
```
