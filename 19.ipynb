На вход программе подается строка текста. Напишите программу, которая выводит на экран символ, который появляется наиболее часто. Если таких символов несколько, следует вывести последний по порядку в строке символ. Следует различать заглавные и строчные буквы, а также буквы русского и английского алфавита.


```python
# Ввод строки текста
text = input("Введите строку текста: ")

# Шаг 1: Создаем словарь для подсчета частоты символов
frequency = {}

# Шаг 2: Подсчитываем частоту каждого символа
for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

# Шаг 3: Находим символ с максимальной частотой, выбирая последний по порядку
max_frequency = -1
most_frequent_char = ''

for char in reversed(text):  # Проходим по строке с конца
    if frequency[char] > max_frequency:
        max_frequency = frequency[char]
        most_frequent_char = char

# Шаг 4: Выводим результат
print("Символ, который встречается наиболее часто (последний по порядку):", most_frequent_char)
```
