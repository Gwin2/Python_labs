Из трёх неизвестных чисел найдите среднее среди них (то есть то, что не самое большое и не самое маленькое). Сделайте это без условных операторов и без сортировки.

```python
def find_middle(a, b, c):
    # Находим максимум и минимум среди трёх чисел
    max_value = max(a, b, c)
    min_value = min(a, b, c)
    
    # Среднее число — это сумма всех чисел минус максимум и минимум
    middle_value = a + b + c - max_value - min_value
    
    return middle_value

# Ввод данных
a = float(input("Введите первое число: "))
b = float(input("Введите второе число: "))
c = float(input("Введите третье число: "))

# Вызов функции и вывод результата
middle = find_middle(a, b, c)
print(f"Среднее число: {middle}")
```
