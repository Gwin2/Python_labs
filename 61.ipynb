Реализуйте функцию is_prime() с использованием генераторных выражений, которая принимает один аргумент:

number — натуральное число.

Функция должна возвращать True, если число number является простым, или False в противном случае.

В задаче удобно воспользоваться функциями all() или any().

```python
def is_prime(number):
    """
    Функция, проверяющая, является ли число number простым.
    """
    if number <= 1:
        return False  # Числа <= 1 не являются простыми
    # Проверяем, что ни одно число от 2 до √number не делит number
    return all(number % x != 0 for x in range(2, int(number**0.5) + 1))


# Пример использования
if __name__ == "__main__":
    # Проверяем несколько чисел
    print(is_prime(2))   # True (простое)
    print(is_prime(3))   # True (простое)
    print(is_prime(4))   # False (не простое)
    print(is_prime(17))  # True (простое)
    print(is_prime(18))  # False (не простое)
```
