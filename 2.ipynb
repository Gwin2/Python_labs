Напишите код, который принимает номер или название месяца вашего рождения и в зависимости от сезона печатает на выходе следующее:
«Вы родились в <НАЗВАНИЕ_МЕСЯЦА>. <ОПИСАНИЕ_СОБЫТИЙ>».

В качестве ОПИСАНИЯ_СОБЫТИЙ будет характеристика сезона: 
- для зимы «За окном падал белый снег»,
- для весны «Птицы пели прекрасные песни»,
- для лета «Солнце светило ярче чем когда-либо»,
- для осени «Урожай был невероятным».

Важно учесть, что пользователи могут ввести любой тип данных в качестве аргумента (не попадитесь на этом и предупредите о том, что «Требуется ввести реальный номер месяца»).

```python
# Функция для получения названия месяца и сезона по номеру или названию месяца
def get_month_and_season(month_input):
    # Словарь для соответствия номера месяца его названию и сезону
    months = {
        1: ("Январь", "зима"),
        2: ("Февраль", "зима"),
        3: ("Март", "весна"),
        4: ("Апрель", "весна"),
        5: ("Май", "весна"),
        6: ("Июнь", "лето"),
        7: ("Июль", "лето"),
        8: ("Август", "лето"),
        9: ("Сентябрь", "осень"),
        10: ("Октябрь", "осень"),
        11: ("Ноябрь", "осень"),
        12: ("Декабрь", "зима")
    }

    # Если введено число (номер месяца)
    if isinstance(month_input, int) and 1 <= month_input <= 12:
        return months.get(month_input)

    # Если введена строка (название месяца)
    elif isinstance(month_input, str):
        month_name = month_input.capitalize()  # Приводим первую букву к верхнему регистру
        for num, (name, season) in months.items():
            if name == month_name:
                return (name, season)

    # Если введен некорректный формат данных
    return None

# Основная часть программы
def main():
    user_input = input("Введите номер или название месяца вашего рождения: ")

    try:
        # Попытка преобразовать ввод в целое число
        month_number = int(user_input)
        result = get_month_and_season(month_number)
    except ValueError:
        # Если не удалось преобразовать в число, считаем это строкой
        result = get_month_and_season(user_input)

    # Проверяем результат
    if result is None:
        print("Требуется ввести реальный номер месяца (1-12) или корректное название месяца.")
    else:
        month_name, season = result

        # Словарь с описанием событий для каждого сезона
        descriptions = {
            "зима": "За окном падал белый снег",
            "весна": "Птицы пели прекрасные песни",
            "лето": "Солнце светило ярче чем когда-либо",
            "осень": "Урожай был невероятным"
        }

        description = descriptions.get(season, "Неизвестное описание")
        print(f"Вы родились в {month_name}. {description}")

# Запуск программы
if __name__ == "__main__":
    main()
```
