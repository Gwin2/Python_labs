Напишите программу, которая принимает на входе ваши имя, возраст, родной город и выводит строку:

“Добрый день, (имя)! Вам (возраст). Вы из города (город).”

```python
# Ввод данных пользователем
name = input("Введите ваше имя: ")
age = input("Введите ваш возраст: ")
city = input("Введите ваш родной город: ")

# Формирование и вывод строки
output = f"Добрый день, {name}! Вам {age}. Вы из города {city}."
print(output)
```
